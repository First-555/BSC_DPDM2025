```python
# ตรวจสอบและแปลงคอลัมน์ 'measure_datetime' ให้เป็น datetime object หากยังไม่ได้แปลง
if not pd.api.types.is_datetime64_any_dtype(filtered_Dailyrain['measure_datetime']):
    filtered_Dailyrain['measure_datetime'] = pd.to_datetime(filtered_Dailyrain['measure_datetime'])

# Group by station and find missing dates
missing_records_summary = {}

for station_code, group in filtered_Dailyrain.groupby('station_code'):
    # Get the min and max date for the current station
    min_date = group['measure_datetime'].min()
    max_date = group['measure_datetime'].max()

    # Create a complete date range for this station (inclusive of min_date and max_date)
    full_date_range = pd.date_range(start=min_date.normalize(), end=max_date.normalize(), freq='D')
    total_expected_days = len(full_date_range)

    # Get the unique dates present in the data for this station
    present_dates = group['measure_datetime'].dt.normalize().unique()

    # Find the missing dates
    missing_dates = full_date_range[~full_date_range.isin(present_dates)]
    missing_count = len(missing_dates)

    if missing_count > 0:
        missing_proportion = (missing_count / total_expected_days) * 100
        missing_records_summary[station_code] = {
            'count': missing_count,
            'proportion': f"{missing_proportion:.2f}%",
            'example_missing_dates': missing_dates.tolist()[:5] # Show up to 5 example missing dates
        }

if missing_records_summary:
    print("สรุปข้อมูล Record ที่หายไปของแต่ละสถานี:")
    # Sort by the number of missing records for better readability
    sorted_summary = sorted(missing_records_summary.items(), key=lambda item: item[1]['count'], reverse=True)
    for station, summary in sorted_summary:
        print(f"- สถานี {station}: ข้อมูลหายไป {summary['count']} วัน ({summary['proportion']})")
        if summary['example_missing_dates']:
            print(f"  ตัวอย่างวันที่หายไป: {', '.join([d.strftime('%Y-%m-%d') for d in summary['example_missing_dates']])}")
else:
    print("ไม่พบข้อมูล Record ที่หายไปในสถานีใดๆ")
```

```python
# Make a copy to work with, to avoid modifying the original filtered_Dailyrain directly
filtered_Dailyrain_imputed = filtered_Dailyrain.copy()

# Extract Day of Year for grouping
filtered_Dailyrain_imputed['day_of_year'] = filtered_Dailyrain_imputed['measure_datetime'].dt.dayofyear

# Calculate median 'data' for each station_code and day_of_year
median_rainfall_by_day = filtered_Dailyrain_imputed.groupby(['station_code', 'day_of_year'])['data'].median().reset_index()
median_rainfall_by_day.rename(columns={'data': 'median_data'}, inplace=True)

# Merge the median values back into the main DataFrame
filtered_Dailyrain_imputed = pd.merge(
    filtered_Dailyrain_imputed,
    median_rainfall_by_day,
    on=['station_code', 'day_of_year'],
    how='left'
)

# Fill NaN values in 'data' column with 'median_data'
filtered_Dailyrain_imputed['data'] = filtered_Dailyrain_imputed['data'].fillna(filtered_Dailyrain_imputed['median_data'])

# Drop the temporary 'day_of_year' and 'median_data' columns
filtered_Dailyrain_imputed.drop(columns=['day_of_year', 'median_data'], inplace=True)

print("แสดงข้อมูล 5 แถวแรกหลังจากเติมค่าที่หายไปด้วยค่ามัธยฐาน:")
display(filtered_Dailyrain_imputed.head())

```

```python
print("ตรวจสอบว่ายังมีค่าที่หายไปในคอลัมน์ 'data' หรือไม่หลังจากเติม:")
print(filtered_Dailyrain_imputed['data'].isnull().sum())
```

จากการดำเนินการแทนค่าข้อมูลที่หายไปในตัวแปรวันเดือนปี พบว่าการเลือกใช้ค่า มัธยฐาน (median) มีความเหมาะสมมากกว่าการใช้ค่าเฉลี่ย (mean) เนื่องจากลักษณะของข้อมูลวันเดือนปีมีการกระจายที่ไม่สมมาตรและมีความเป็นไปได้ที่จะพบค่าผิดปกติ (outliers) เช่น วันที่ที่เก่าหรือใหม่ผิดจากช่วงเวลาหลักของข้อมูล ซึ่งค่าเฉลี่ยมีความไวต่อค่าผิดปกติดังกล่าว อาจส่งผลให้ค่าที่นำมาแทนไม่สะท้อนค่ากลางที่แท้จริงของข้อมูล

ในขณะที่ค่า median เป็นค่ากลางตามลำดับของข้อมูล จึงไม่ได้รับอิทธิพลจากค่าที่มีความแตกต่างอย่างมาก ทำให้ค่าที่แทนลงไปมีความใกล้เคียงกับแนวโน้มของข้อมูลเดิมมากกว่า นอกจากนี้ ข้อมูลวันเดือนปีจัดเป็นข้อมูลเชิงลำดับตามเวลา การใช้ median จึงสามารถแทนค่ากลางของช่วงเวลาได้อย่างเหมาะสม และช่วยรักษาโครงสร้างการกระจายของข้อมูลไว้ได้

ดังนั้น การใช้ค่า median ในการแทนค่าที่หายไปของข้อมูลวันเดือนปีจึงช่วยลดความบิดเบือนของข้อมูลหลังการประมวลผล และเพิ่มความน่าเชื่อถือของผลการวิเคราะห์ในขั้นตอนถัดไป เมื่อเปรียบเทียบกับการใช้ค่า mean ซึ่งอาจทำให้ค่าที่แทนมีความคลาดเคลื่อนจากสภาพความเป็นจริงของข้อมูล
